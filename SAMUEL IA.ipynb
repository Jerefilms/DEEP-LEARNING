```python
# CLUSTERING NO SUPERVISADO CON K-MEANS SAMUEL DAVID CELIS


import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans


X, _ = make_blobs(n_samples=400, centers=4, cluster_std=0.68, random_state=42)

kmeans = KMeans(n_clusters=4, n_init=10, random_state=42)

kmeans.fit(X)


y_kmeans = kmeans.predict(X)
centroides = kmeans.cluster_centers_


plt.figure(figsize=(9, 6))


plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis', alpha=0.7, label='Puntos agrupados')


plt.scatter(centroides[:, 0], centroides[:, 1], c='red', s=250, marker='X', label='Centroides')

plt.title('Ejemplo Práctico de Machine Learning No Supervisado: K-Means', fontsize=12)
plt.xlabel('Variable / Característica X1')
plt.ylabel('Variable / Característica X2')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

print("Coordenadas de los centroides finales encontrados:")
print(centroides)
```
