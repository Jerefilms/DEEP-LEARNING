```python
# Aca importamos las librerias
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, metrics
import matplotlib.pyplot as plt


# Fijamos las semillas aleatorias para garantizar que los resultados sean reproducibles
np.random.seed(42)
tf.random.set_seed(42)

print(f"Versión de TensorFlow activa en Colab: {tf.__version__}")
print("Verificando aceleración por GPU:", tf.config.list_physical_devices('GPU'))

```

```python

# Generamos muestras que simulan imágenes de rostros autorizados preprocesados
# Dimensiones de imagen: 105x105 píxeles con 3 canales de color (RGB), normalizados en [0, 1]

def generate_dummy_triplets(num_samples=300):
    """
    Crea tripletas compuestas por:
    - Ancla (Anchor): Rostro de referencia.
    - Positiva (Positive): Mismo rostro con ligeras variaciones de iluminación/ruido.
    - Negativa (Negative): Rostro de una identidad totalmente diferente.
    """
    # Anclas
    anchors = np.random.rand(num_samples, 105, 105, 3).astype(np.float32)

    # Positivas
    positives = anchors + np.random.normal(0, 0.05, anchors.shape).astype(np.float32)
    positives = np.clip(positives, 0.0, 1.0) # Aseguramos rango válido de píxeles [0, 1]

    # Negativas
    negatives = np.random.rand(num_samples, 105, 105, 3).astype(np.float32)

    return anchors, positives, negatives

# Cargar el dataset de pruebas
anchors, positives, negatives = generate_dummy_triplets(300)
print(f"Dataset sintético generado: {anchors.shape[0]} tripletas preparadas para entrenamiento.")

```

```python
# creamos la subred
def build_embedding_network(input_shape=(105, 105, 3)):
    """
    Construye la red neuronal convolucional encargada de proyectar la imagen
    de un rostro a un vector de características (embedding) de 128 dimensiones.
    """
    model = models.Sequential(name="Embedding_Network")

    # Extracción de bordes y texturas primarias
    model.add(layers.Conv2D(64, (10, 10), activation='relu', input_shape=input_shape))
    model.add(layers.MaxPooling2D((2, 2)))

    #Formas intermedias
    model.add(layers.Conv2D(128, (7, 7), activation='relu'))
    model.add(layers.MaxPooling2D((2, 2)))

    #Rasgos faciales complejos
    model.add(layers.Conv2D(128, (4, 4), activation='relu'))
    model.add(layers.MaxPooling2D((2, 2)))

    #Abstracción de alto nivel
    model.add(layers.Conv2D(256, (4, 4), activation='relu'))
    model.add(layers.Flatten())

    #Vector de Embeddig de 128 dimensiones
    model.add(layers.Dense(128, activation=None))

    #Ajusta la magnitud del vector a 1
    model.add(layers.Lambda(lambda x: tf.math.l2_normalize(x, axis=1)))

    return model

# Resumen técnico de la subred convolucional
embedding_net = build_embedding_network()
embedding_net.summary()

```

```python

# Definimos los tres puertos de entrada
input_anchor = layers.Input(shape=(105, 105, 3), name="Anchor_Input")
input_positive = layers.Input(shape=(105, 105, 3), name="Positive_Input")
input_negative = layers.Input(shape=(105, 105, 3), name="Negative_Input")

# LAS TRES ENTRADAS PASAN POR LA MISMA SUBRED
emb_anchor = embedding_net(input_anchor)
emb_positive = embedding_net(input_positive)
emb_negative = embedding_net(input_negative)

# Concatenamos los tres embeddings
output_triplet = layers.concatenate([emb_anchor, emb_positive, emb_negative], axis=1)

# Construcción del modelo final siamés
siamese_network = models.Model(
    inputs=[input_anchor, input_positive, input_negative],
    outputs=output_triplet,
    name="Siamese_Triplet_Network"
)

```

```python
#hiperloss
def custom_triplet_loss(y_true, y_pred, margin=0.2):
    """
    Función de pérdida Triplet Loss:
    L = max(0, ||emb(A) - emb(P)||^2 - ||emb(A) - emb(N)||^2 + margin)
    """
    # Desempacamos los tres vectores de embedding
    anchor = y_pred[:, :128]
    positive = y_pred[:, 128:256]
    negative = y_pred[:, 256:]

    # Calculamos la distancia euclidiana al cuadrado para el par positivo
    pos_dist = tf.reduce_sum(tf.square(anchor - positive), axis=-1)

    # alculamos la distancia euclidiana al cuadrado para el par negativo
    neg_dist = tf.reduce_sum(tf.square(anchor - negative), axis=-1)

    #Aplicamos el margen de seguridad
    basic_loss = pos_dist - neg_dist + margin

    #Retornamos el promedio del error sobre el lote
    return tf.reduce_mean(tf.maximum(basic_loss, 0.0))

#Compilación del modelo siamés con optimizador Adam
siamese_network.compile(
    optimizer=optimizers.Adam(learning_rate=0.0001),
    loss=custom_triplet_loss
)

```

```python

# Creamos un tensor dummy de ceros
y_dummy = np.zeros((anchors.shape[0], 384))

print("\nIniciando el entrenamiento de la Red Siamesa...")
history = siamese_network.fit(
    x=[anchors, positives, negatives],
    y=y_dummy,
    batch_size=16,
    epochs=10,
    verbose=1
)


def verify_faces(img1, img2, threshold=0.5):
    """
    Recibe dos imágenes, extrae sus embeddings y determina si pertenecen a la misma persona.
    """
    # Añadimos dimensión de batch (1, 105, 105, 3) y predecimos el embedding
    emb1 = embedding_net.predict(np.expand_dims(img1, axis=0), verbose=0)
    emb2 = embedding_net.predict(np.expand_dims(img2, axis=0), verbose=0)

    # Calculamos la distancia euclidiana
    distance = np.linalg.norm(emb1 - emb2)
    is_same = distance < threshold
    return distance, is_same

# Evaluación con una muestra de prueba
test_anchor = anchors[0]
test_pos = positives[0]
test_neg = negatives[0]

dist_pos, match_pos = verify_faces(test_anchor, test_pos)
dist_neg, match_neg = verify_faces(test_anchor, test_neg)

print("\n==================================================")
print("     RESULTADOS DE VERIFICACIÓN FACIAL (ONE-SHOT)   ")
print("==================================================")
print(f"Par Positivo (Misma Persona) -> Distancia: {dist_pos:.4f} | ¿Verificado?: {match_pos}")
print(f"Par Negativo (Persona Distinta) -> Distancia: {dist_neg:.4f} | ¿Verificado?: {match_neg}")
print("==================================================")
```
