```python
#Samuel David Celis Jeremiche


#Instalación
!pip install -q transformers torch matplotlib networkx

import torch
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from transformers import pipeline, AutoTokenizer, AutoModel

print("Entorno configurado correctamente.\n")


# MODELOS TRANSFORMERS (HUGGING FACE)


print("--- 1. ANÁLISIS DE SENTIMIENTOS (Arquitectura Encoder - DistilBERT) ---")
clasificador = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
textos_prueba = [
    "Transformers revolutionized natural language processing with self-attention mechanisms.",
    "The delivery was delayed and the customer service response was unhelpful."
]
for texto in textos_prueba:
    resultado = clasificador(texto)[0]
    print(f"Texto: '{texto}'")
    print(f" -> Predicción: {resultado['label']} (Confianza: {resultado['score']:.4f})\n")

print("--- 2. PREDICCIÓN DE CONTEXTO / FILL-MASK (Arquitectura Encoder - BERT) ---")
enmascarador = pipeline("fill-mask", model="bert-base-uncased")
oracion_mask = "The self-attention mechanism allows the model to focus on relevant [MASK] in a sentence."
predicciones = enmascarador(oracion_mask)[:3]
print(f"Oración: '{oracion_mask}'")
for pred in predicciones:
    print(f" -> Candidato: '{pred['token_str']}' (Score: {pred['score']:.4f})")
print()

print("--- 3. GENERACIÓN DE TEXTO (Arquitectura Decoder - GPT-2) ---")
generador = pipeline("text-generation", model="gpt2")
prompt = "Artificial intelligence and Transformer models are transforming modern healthcare by"
salida_gen = generador(prompt, max_length=45, num_return_sequences=1, truncation=True)[0]['generated_text']
print(f"Prompt inicial: '{prompt}'")
print(f"Texto generado:\n -> {salida_gen}\n")

#
# VISUALIZACIÓN DEL MECANISMO DE AUTOATENCIÓN

print("--- 4. EXTRACCIÓN Y VISUALIZACIÓN DE MATRIZ DE ATENCIÓN ---")
modelo_nombre = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(modelo_nombre)
model = AutoModel.from_pretrained(modelo_nombre, output_attentions=True)

frase_atencion = "The cat did not cross the street because it was tired"
inputs = tokenizer(frase_atencion, return_tensors="pt")
outputs = model(**inputs)

# Extraer atención de la última capa, promediando todas las cabezas (Multi-Head)
atencion = outputs.attentions[-1][0].mean(dim=0).detach().numpy()
tokens = tokenizer.convert_ids_to_tokens(inputs["input_ids"][0])

# Graficar mapa de calor de autoatención
plt.figure(figsize=(8, 6))
plt.imshow(atencion, cmap="Blues")
plt.xticks(range(len(tokens)), tokens, rotation=45, ha="right")
plt.yticks(range(len(tokens)), tokens)
plt.colorbar(label="Peso de Atención (Softmax)")
plt.title("Mapa de Autoatención (Self-Attention) - Capa Final BERT")
plt.tight_layout()
plt.savefig("matriz_atencion_transformer.png", dpi=300)
plt.show()
print("[GUARDADO] Gráfico de atención guardado como 'matriz_atencion_transformer.png'\n")


#GENERACIÓN AUTOMÁTICA DEL MAPA MENTAL DE TRANSFORMERS

print("--- 5. GENERANDO MAPA MENTAL JERÁRQUICO DE MODELOS TRANSFORMERS ---")

G = nx.DiGraph()

# Nodo Central
centro = "MODELOS\nTRANSFORMERS\n(Vaswani et al., 2017)"

# Ramas Principales
ramas = {
    "1. Fundamentos": ["Sin recurrencia (No RNN)", "Procesamiento paralelo", "Positional Encoding"],
    "2. Mecanismo de Atención": ["Self-Attention", "Query, Key, Value (Q,K,V)", "Multi-Head Attention"],
    "3. Arquitectura": ["Encoder (Comprensión)", "Decoder (Generación)", "Feed-Forward & LayerNorm"],
    "4. Aplicaciones NLP": ["Traducción automática", "Análisis de sentimientos", "Modelos LLM (BERT / GPT)"]
}

# Construir nodos y conexiones
G.add_node(centro, nivel=0)
for rama, subnodos in ramas.items():
    G.add_node(rama, nivel=1)
    G.add_edge(centro, rama)
    for sub in subnodos:
        G.add_node(sub, nivel=2)
        G.add_edge(rama, sub)

# Posicionamiento radial estructurado
pos = {centro: np.array([0.0, 0.0])}
angulos_ramas = [np.pi * 0.75, np.pi * 0.25, -np.pi * 0.75, -np.pi * 0.25]
radio_1 = 2.2
radio_2 = 4.3

for idx, (rama, subnodos) in enumerate(ramas.items()):
    ang = angulos_ramas[idx]
    pos[rama] = np.array([radio_1 * np.cos(ang), radio_1 * np.sin(ang)])

    # Distribuir subnodos en abanico alrededor de su rama padre
    offsets = [-0.28, 0.0, 0.28]
    for s_idx, sub in enumerate(subnodos):
        sub_ang = ang + offsets[s_idx]
        pos[sub] = np.array([radio_2 * np.cos(sub_ang), radio_2 * np.sin(sub_ang)])

# Dibujar el Mapa Mental
plt.figure(figsize=(14, 9))
ax = plt.gca()

# Dibujar aristas curvas suaves
nx.draw_networkx_edges(
    G, pos, ax=ax, edge_color="#7F8C8D", width=2.0,
    arrows=True, arrowsize=15, connectionstyle="arc3,rad=0.08"
)

# Estilos por nivel jerárquico
colores_nivel = {0: "#1F497D", 1: "#2E75B6", 2: "#EAF2F8"}
text_colores = {0: "white", 1: "white", 2: "#1B2631"}
tamanos_fuente = {0: 11, 1: 9.5, 2: 8.5}

for nodo, (x, y) in pos.items():
    nivel = G.nodes[nodo]["nivel"]
    bbox_props = dict(
        boxstyle="round,pad=0.6" if nivel < 2 else "round,pad=0.45",
        fc=colores_nivel[nivel],
        ec="#1F497D",
        lw=1.8 if nivel < 2 else 1.0
    )
    ax.text(
        x, y, nodo, ha="center", va="center",
        fontsize=tamanos_fuente[nivel],
        fontweight="bold" if nivel < 2 else "normal",
        color=text_colores[nivel],
        bbox=bbox_props
    )

plt.title("Mapa Mental: Arquitectura, Mecanismos y Aplicaciones de los Modelos Transformers", fontsize=14, fontweight="bold", pad=15)
plt.xlim(-5.8, 5.8)
plt.ylim(-3.8, 3.8)
plt.axis("off")
plt.tight_layout()
plt.savefig("mapa_mental_transformers.png", dpi=300)
plt.show()
print("[GUARDADO] Mapa Mental exportado en alta calidad como 'mapa_mental_transformers.png'")
```
