```python
# ==============================================================================
# SCRIPT DE GOOGLE COLAB CON VISUALIZACIÓN DIRECTA EN PANTALLA
# ==============================================================================

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from IPython.display import display, Image

print("Sube tus imágenes originales (referencia.jpg, oscura.jpg, desenfocada.jpg, fondo.jpg, angulo.jpg):")
uploaded = files.upload()

# Cargar imágenes al diccionario de trabajo
imagenes = {}
for filename in uploaded.keys():
    img = cv2.imread(filename)
    if img is not None:
        imagenes[filename] = img
        print(f"Cargada correctamente: {filename}")

# ------------------------------------------------------------------------------
# 1. FIGURA 3: Recortes y Análisis de Nitidez (Laplaciano)
# ------------------------------------------------------------------------------
if 'referencia.jpg' in imagenes and 'desenfocada.jpg' in imagenes:
    h, w, _ = imagenes['referencia.jpg'].shape
    y0, y1 = int(h * 0.1), int(h * 0.9)
    x0, x1 = int(w * 0.2), int(w * 0.8)

    ref_roi = cv2.cvtColor(imagenes['referencia.jpg'][y0:y1, x0:x1], cv2.COLOR_BGR2GRAY)
    des_roi = cv2.cvtColor(imagenes['desenfocada.jpg'][y0:y1, x0:x1], cv2.COLOR_BGR2GRAY)

    lap_ref = cv2.Laplacian(ref_roi, cv2.CV_64F)
    var_ref = lap_ref.var()
    lap_ref_abs = np.uint8(np.absolute(lap_ref))

    lap_des = cv2.Laplacian(des_roi, cv2.CV_64F)
    var_des = lap_des.var()
    lap_des_abs = np.uint8(np.absolute(lap_des))

    fig, axes = plt.subplots(2, 2, figsize=(8, 10))
    axes[0, 0].imshow(cv2.cvtColor(imagenes['referencia.jpg'][y0:y1, x0:x1], cv2.COLOR_BGR2RGB))
    axes[0, 0].set_title("Referencia: recorte de la figura")
    axes[0, 0].axis('off')

    axes[0, 1].imshow(cv2.cvtColor(imagenes['desenfocada.jpg'][y0:y1, x0:x1], cv2.COLOR_BGR2RGB))
    axes[0, 1].set_title("Desenfocada: recorte de la figura")
    axes[0, 1].axis('off')

    axes[1, 0].imshow(lap_ref_abs, cmap='gray')
    axes[1, 0].set_title(f"|Laplaciano| (varianza = {var_ref:.1f})")
    axes[1, 0].axis('off')

    axes[1, 1].imshow(lap_des_abs, cmap='gray')
    axes[1, 1].set_title(f"|Laplaciano| (varianza = {var_des:.1f})")
    axes[1, 1].axis('off')

    plt.tight_layout()
    plt.savefig("figura3_nitidez.png", dpi=300)
    plt.show() # Muestra la gráfica en pantalla

# ------------------------------------------------------------------------------
# 2. FIGURA 5: Comparación de los 4 Métodos de Tratamiento
# ------------------------------------------------------------------------------
if 'oscura.jpg' in imagenes:
    img_dark = imagenes['oscura.jpg']

    # Gamma 0.44 en canal V de HSV
    hsv = cv2.cvtColor(img_dark, cv2.COLOR_BGR2HSV)
    lut_gamma = np.array([((i / 255.0) ** 0.44) * 255 for i in range(256)]).astype(np.uint8)
    hsv[:, :, 2] = cv2.LUT(hsv[:, :, 2], lut_gamma)
    res_gamma = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)

    # Ajuste Lineal
    res_lineal = cv2.convertScaleAbs(img_dark, alpha=2.0, beta=40)

    # Ecualización Global
    hsv2 = cv2.cvtColor(img_dark, cv2.COLOR_BGR2HSV)
    hsv2[:, :, 2] = cv2.equalizeHist(hsv2[:, :, 2])
    res_eq = cv2.cvtColor(hsv2, cv2.COLOR_HSV2BGR)

    # CLAHE
    lab = cv2.cvtColor(img_dark, cv2.COLOR_BGR2Lab)
    clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8, 8))
    lab[:, :, 0] = clahe.apply(lab[:, :, 0])
    res_clahe = cv2.cvtColor(lab, cv2.COLOR_Lab2BGR)

    fig, axes = plt.subplots(1, 4, figsize=(16, 5))
    methods = [
        ("Gamma 0.44 (elegida)", res_gamma),
        ("Lineal (a=2, b=40)", res_lineal),
        ("Ecualización (V)", res_eq),
        ("CLAHE (L)", res_clahe)
    ]

    rh, rw, _ = img_dark.shape
    y_min, y_max = int(rh * 0.1), int(rh * 0.9)
    x_min, x_max = int(rw * 0.2), int(rw * 0.8)

    for i, (name, img_res) in enumerate(methods):
        crop = img_res[y_min:y_max, x_min:x_max]
        axes[i].imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
        axes[i].set_title(name)
        axes[i].axis('off')

    plt.tight_layout()
    plt.savefig("figura5_metodos.png", dpi=300)
    plt.show() # Muestra la gráfica en pantalla

# ------------------------------------------------------------------------------
# 3. FIGURA 6: Antes, Después y Detalles de la Corrección Gamma
# ------------------------------------------------------------------------------
if 'oscura.jpg' in imagenes and 'res_gamma' in locals():
    fig, axes = plt.subplots(2, 2, figsize=(10, 10))

    axes[0, 0].imshow(cv2.cvtColor(imagenes['oscura.jpg'], cv2.COLOR_BGR2RGB))
    axes[0, 0].set_title("Antes (oscura.jpg)")
    axes[0, 0].axis('off')

    axes[0, 1].imshow(cv2.cvtColor(res_gamma, cv2.COLOR_BGR2RGB))
    axes[0, 1].set_title("Después (gamma = 0.44)")
    axes[0, 1].axis('off')

    det_antes = imagenes['oscura.jpg'][y_min:y_max, x_min:x_max]
    axes[1, 0].imshow(cv2.cvtColor(det_antes, cv2.COLOR_BGR2RGB))
    axes[1, 0].set_title("Detalle de la figura")
    axes[1, 0].axis('off')

    det_despues = res_gamma[y_min:y_max, x_min:x_max]
    axes[1, 1].imshow(cv2.cvtColor(det_despues, cv2.COLOR_BGR2RGB))
    axes[1, 1].set_title("Detalle de la figura")
    axes[1, 1].axis('off')

    plt.tight_layout()
    plt.savefig("figura6_antes_despues.png", dpi=300)
    plt.show() # Muestra la gráfica en pantalla

    # Forzar visualización directa interactiva usando IPython
    print("\nVisualización directa de la imagen tratada final:")
    display(Image('figura6_antes_despues.png'))
```
